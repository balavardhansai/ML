```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = "/kaggle/working/final_dataset/dataset/posts_short.csv"

df = pd.read_csv(file_path, low_memory=False)

print("Dataset loaded:", df.shape)
```

```python
questions = df[df["post_type_id"] == 1].copy()

def count_tags(tags):
    if pd.isna(tags) or str(tags).strip() == "":
        return 0
    return len(str(tags).strip("<>").split("><"))

questions["tag_count"] = questions["tags"].apply(count_tags)

print("Total questions:", len(questions))
display(questions[["id", "tags", "tag_count"]].head(10))
```

**Question-5:Determine the ratio of duplicate questions asked in each month**

```python
import pandas as pd
import matplotlib.pyplot as plt

base = "/kaggle/working/final_dataset/dataset"

q = pd.read_csv(f"{base}/posts_short.csv",
                usecols=["id", "post_type_id", "creation_date"])
q["creation_date"] = pd.to_datetime(q["creation_date"], errors="coerce")
q = q[(q["post_type_id"] == 1) &
      (q["creation_date"].dt.year.between(2008, 2018))].copy()

links = pd.read_csv(f"{base}/postLinks.csv",
                    usecols=["post_id", "link_type_id"])
dup_ids = links.loc[links["link_type_id"] == 3, "post_id"].unique()

q["is_dup"] = q["id"].isin(dup_ids).astype(int)
q["month"] = q["creation_date"].dt.to_period("M")

monthly = q.groupby("month").agg(
    dup=("is_dup", "sum"), total=("id", "count")
)
monthly["ratio"] = monthly["dup"] / monthly["total"] * 100
monthly.index = monthly.index.to_timestamp()

monthly = monthly.reindex(
    pd.date_range("2008-01-01", "2018-12-01", freq="MS")
)

plt.figure(figsize=(12, 5))
plt.plot(monthly.index, monthly["ratio"])
plt.title("Monthly Duplicate Question Ratio (2008–2018)")
plt.xlabel("Month")
plt.ylabel("Duplicate Questions (%)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

***Question-6:Determine the percentage of duplicate questions associated with different tags***

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

base = "/kaggle/working/final_dataset/dataset"

q = pd.read_csv(
    f"{base}/posts_short.csv",
    usecols=["id", "post_type_id", "tags"]
)
links = pd.read_csv(
    f"{base}/postLinks.csv",
    usecols=["post_id", "link_type_id"]
)

dup_ids = links.loc[links["link_type_id"] == 3, "post_id"].dropna().unique()
q = q[q["post_type_id"] == 1].copy()
q["is_dup"] = q["id"].isin(dup_ids)

total, duplicates = Counter(), Counter()

for tags, is_dup in zip(q["tags"].dropna(), q.loc[q["tags"].notna(), "is_dup"]):
    tag_list = str(tags).strip("<>").split("><")
    total.update(tag_list)
    if is_dup:
        duplicates.update(tag_list)

result = pd.DataFrame([
    {"Tag": tag, "Total": count,
     "Duplicate_Percentage": duplicates[tag] / count * 100}
    for tag, count in total.items() if count > 0
])

result = result.sort_values("Duplicate_Percentage", ascending=False).head(20)
result = result.sort_values("Duplicate_Percentage")

plt.figure(figsize=(11, 8))
plt.barh(result["Tag"], result["Duplicate_Percentage"])
plt.title("Top 20 Tags by Duplicate Question Percentage")
plt.xlabel("Duplicate Questions (%)")
plt.ylabel("Tag")
plt.grid(axis="x", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()

display(result.sort_values("Duplicate_Percentage", ascending=False))
```

***Question-7:Determine the time required to close duplicate questions***

```python
import pandas as pd
import matplotlib.pyplot as plt

base = "/kaggle/working/final_dataset/dataset"

links = pd.read_csv(
    f"{base}/postLinks.csv",
    usecols=["creation_date", "post_id", "link_type_id"]
)
links = links[links["link_type_id"] == 3].copy()
links["creation_date"] = pd.to_datetime(links["creation_date"], errors="coerce")

posts = pd.read_csv(
    f"{base}/posts_short.csv",
    usecols=["id", "creation_date"]
)
posts["creation_date"] = pd.to_datetime(posts["creation_date"], errors="coerce")

data = links.merge(posts, left_on="post_id", right_on="id", suffixes=("_closed", "_asked"))
data["days"] = (
    data["creation_date_closed"] - data["creation_date_asked"]
).dt.total_seconds() / 86400

data = data[data["days"].ge(0) & data["days"].notna()]

plt.figure(figsize=(10, 5))
plt.hist(data["days"], bins=30, edgecolor="black")
plt.title("Time Required to Close Duplicate Questions")
plt.xlabel("Days")
plt.ylabel("Number of Duplicate Questions")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

print("Average days:", round(data["days"].mean(), 2))
```

***Question-8:Distribution of the reputation of users whose questions are closed as
duplicates***

```python
import pandas as pd
import matplotlib.pyplot as plt

base = "/kaggle/working/final_dataset/dataset"

links = pd.read_csv(
    f"{base}/postLinks.csv",
    usecols=["post_id", "link_type_id"]
)
dup_ids = links.loc[links["link_type_id"] == 3, "post_id"].unique()

posts = pd.read_csv(
    f"{base}/posts_short.csv",
    usecols=["id", "post_type_id", "owner_user_id"]
)
users = pd.read_csv(
    f"{base}/users.csv",
    usecols=["id", "reputation"]
)

dup_questions = posts[
    (posts["post_type_id"] == 1) & posts["id"].isin(dup_ids)
]
data = dup_questions.merge(
    users, left_on="owner_user_id", right_on="id", how="left"
).dropna(subset=["reputation"])

data["Reputation Category"] = pd.cut(
    data["reputation"],
    bins=[-1, 100, 1000, 10000, float("inf")],
    labels=["0–100", "101–1,000", "1,001–10,000", "10,000+"]
)

data["Reputation Category"].value_counts().sort_index().plot(
    kind="bar", figsize=(9, 5), edgecolor="black"
)
plt.title("Reputation of Users Asking Duplicate Questions")
plt.xlabel("Reputation Category")
plt.ylabel("Number of Users' Questions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = "/kaggle/working/final_dataset/Arjun_Assignment_data/dataset-20210607T020316Z-001/dataset/posts_short.csv"

# Load question data
df = pd.read_csv(file_path, usecols=["id", "post_type_id", "score"])
questions = df[df["post_type_id"] == 1].copy()
questions["score"] = pd.to_numeric(questions["score"], errors="coerce")
questions = questions.dropna(subset=["score"])

# Categorize question scores
questions["Score Category"] = questions["score"].apply(
    lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"
)

# Calculate counts and percentages
category_order = ["Positive", "Zero", "Negative"]
counts = questions["Score Category"].value_counts().reindex(category_order, fill_value=0)
percentages = counts / len(questions) * 100

results = pd.DataFrame({
    "Number of Questions": counts,
    "Percentage (%)": percentages.round(2)
})

print("QUESTION SCORE DISTRIBUTION")
print("===========================")
print("Total questions analysed:", len(questions))
print(results)

# Create bar chart
plt.figure(figsize=(9, 5))
bars = plt.bar(counts.index, counts.values)
plt.bar_label(bars, labels=[f"{v:,}" for v in counts.values], padding=3)
plt.title("Distribution of Stack Overflow Questions by Score")
plt.xlabel("Score Category")
plt.ylabel("Number of Questions")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

# Display the category with the highest count
highest = counts.idxmax()
print(f"\nMost common score category: {highest}")
print(f"Percentage of questions: {percentages[highest]:.2f}%")

print("\nConclusion:")
print(
    f"The {highest.lower()} score category contains the highest number "
    f"of questions, accounting for {percentages[highest]:.2f}% "
    "of the questions analysed."
)
```

***Question-12:Determine the Most Active Users***

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = "/kaggle/working/final_dataset/Arjun_Assignment_data/dataset-20210607T020316Z-001/dataset/posts_short.csv"

# Load the required columns
df = pd.read_csv(
    file_path,
    usecols=["id", "post_type_id", "owner_user_id"]
)

# Select questions only
questions = df[df["post_type_id"] == 1].copy()

# Remove questions without a recorded owner
questions = questions.dropna(subset=["owner_user_id"])

# Count questions asked by each user
user_counts = questions["owner_user_id"].value_counts()

# Select the top 20 users
top_20 = user_counts.head(20).sort_values()

print("TOP 20 MOST ACTIVE USERS")
print("========================")
print(top_20.sort_values(ascending=False).to_string())
print("\nTotal questions analysed:", len(questions))
print("Number of users with recorded IDs:", questions["owner_user_id"].nunique())

# Create horizontal bar chart
plt.figure(figsize=(10, 7))
bars = plt.barh(top_20.index.astype(str), top_20.values)

plt.bar_label(bars, padding=3, fmt="%d")
plt.title("Top 20 Users Who Asked the Most Questions")
plt.xlabel("Number of Questions")
plt.ylabel("User ID")
plt.grid(axis="x", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

# Conclusion
most_active_user = user_counts.index[0]
most_questions = user_counts.iloc[0]

print("\nCONCLUSION")
print(f"User ID {most_active_user:.0f} asked the most questions: {most_questions}.")
print("The bar chart compares question activity among the 20 most active users.")
```

***Question-13:Determine the Relationship Between Tags and Question Scores***

```python
import pandas as pd
import matplotlib.pyplot as plt

path = "/kaggle/working/final_dataset/Arjun_Assignment_data/dataset-20210607T020316Z-001/dataset/posts_short.csv"

df = pd.read_csv(path, usecols=["post_type_id", "score", "tags"])
df = df[df["post_type_id"] == 1].dropna(subset=["score", "tags"])

df["tags"] = df["tags"].str.findall(r"<([^<>]+)>")
data = df.explode("tags")

stats = data.groupby("tags")["score"].agg(["mean", "count"])
top15 = stats[stats["count"] >= 100].nlargest(15, "mean").sort_values("mean")

print(top15.round(2))

top15["mean"].plot(kind="barh", figsize=(8, 5), title="Top 15 Tags by Average Score")
plt.xlabel("Average Question Score")
plt.ylabel("Tag")
plt.tight_layout()
plt.show()
```

***Question-14:Determine How Question Activity Changes Over the Years***

```python
import pandas as pd
import matplotlib.pyplot as plt

path = "/kaggle/working/final_dataset/Arjun_Assignment_data/dataset-20210607T020316Z-001/dataset/posts_short.csv"

df = pd.read_csv(path, usecols=["post_type_id", "creation_date"])
df = df[df["post_type_id"] == 1].copy()
df["creation_date"] = pd.to_datetime(df["creation_date"], errors="coerce")
df = df.dropna(subset=["creation_date"])
df = df[df["creation_date"].dt.year.between(2008, 2018)]

yearly = df["creation_date"].dt.year.value_counts().sort_index()

print("Questions asked each year:")
print(yearly)

# Output 1: Line graph
yearly.plot(kind="line", marker="o", figsize=(9, 4), title="Yearly Question Trends")
plt.xlabel("Year")
plt.ylabel("Number of Questions")
plt.xticks(range(2008, 2019))
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

# Output 2: Bar chart
yearly.plot(kind="bar", figsize=(9, 4), title="Questions Asked Each Year")
plt.xlabel("Year")
plt.ylabel("Number of Questions")
plt.tight_layout()
plt.show()

# Output 3: Highest activity
print("\nYear with the most questions:", yearly.idxmax())
print("Number of questions:", yearly.max())
```

***Question-15:What is the relationship between question score and the number of tags***

```python

import pandas as pd
import matplotlib.pyplot as plt

path = "/kaggle/working/final_dataset/Arjun_Assignment_data/dataset-20210607T020316Z-001/dataset/posts_short.csv"

df = pd.read_csv(path, usecols=["post_type_id", "score", "tags"])
df = df[df["post_type_id"] == 1].dropna(subset=["score", "tags"])

# Use a sample for faster processing
df = df.sample(n=min(100000, len(df)), random_state=42).copy()

# Count tags quickly
df["Number of Tags"] = df["tags"].str.count("<", flags=0)

# Output 1: Average score
result = df.groupby("Number of Tags")["score"].mean().round(2)
print("1. Average score by number of tags:")
print(result)

# Output 2: Number of questions
print("\n2. Number of questions by tag count:")
print(df["Number of Tags"].value_counts().sort_index())

# Output 3: Line graph
result.plot(kind="line", marker="o", figsize=(8, 4),
            title="Average Score vs Number of Tags")
plt.xlabel("Number of Tags")
plt.ylabel("Average Score")
plt.grid(True)
plt.tight_layout()
plt.show()

# Output 4: Scatter plot
plt.scatter(df["Number of Tags"], df["score"], alpha=0.3)
plt.title("Question Score vs Number of Tags")
plt.xlabel("Number of Tags")
plt.ylabel("Question Score")
plt.tight_layout()
plt.show()
```
